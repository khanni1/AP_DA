```python
# Q3

'''
NOTE rules of thumb 
Just remember these patterns:
- data.loc[row_label, column_label]
- data.iloc[row_position, column_position]
- data.index[i] gives you the index label of the row at position i.
'''

data = pd.read_csv('hotel_bookings.csv')

set(data['hotel']) # its categotgorical 
# if city hotel then 0 else 0

for i in range(0,len(data['hotel'])):
    if(data.loc[i,'hotel'] == 'City Hotel'):
        data.loc[i,'hotel'] = '1'
    else:
        data.loc[i,'hotel'] = '0'

print(data.sample(10)['hotel'])

            
```
