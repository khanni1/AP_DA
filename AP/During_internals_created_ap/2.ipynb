```python
from urllib.parse import urlparse

url = "https://google.com/search?q=python"

x = urlparse(url)

print(x.scheme)
print(x.netloc)
print(x.path)
print(x.query)
```

```python
import urllib.request as urr

# read a webpage

url = "https://example.com"

response = urr.urlopen(url)

data = response.read()

print(data[:100])

# download a webpage

url = "https://example.com"

urr.urlretrieve(url, "page.html")

print("Downloaded")
```

```python
# basic server → bind → listen → accept and client → connect pattern.

sser = s.socket()

port = 6000
host = '0.0.0.0'

sser.bind((host,port))
sser.listen(1)

c,addr = sser.accept()

print(c,'  ',addr)

msg = 'yalla its working c1'

for i in range(10):
    c.send(msg.encode())


c.close()



```
