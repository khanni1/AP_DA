```python
#  client of 2.ipynbabs

import socket as s

sc = s.socket()

host = '127.0.0.1'
port = 6000

sc.connect((host,port))

msg = sc.recv(1024)

while msg:
    print(msg.decode())
    msg = sc.recv(1024)
    

sc.close()
```
