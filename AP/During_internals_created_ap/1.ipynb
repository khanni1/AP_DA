```python
def regular_method(x, y): 
    n = len(x) 
    x_mean = np.mean(x) 
    y_mean = np.mean(y)
    
    print(f"Mean of x: {x_mean}") 
    print(f"Mean of y: {y_mean}")
    
    # b1 (slope) 
    numerator = np.sum((x - x_mean) * (y - y_mean)) 
    denominator = np.sum((x - x_mean) ** 2) 
    a1 = numerator / denominator
    
    # b0 (intercept)
    a0 = y_mean - a1 * x_mean
    print(a0)
    print(a1)
    
    y_pred = a0 + a1 * x 
    print(f"y_pred: {y_pred}") 
    error = y - y_pred 
    print(f"error: {error}")
    
    plt.scatter(x, y, color="blue", label="Actual Data") 
    plt.plot(x, y_pred, color="red", label="Regression Line") 
    plt.xlabel("x") 
    plt.ylabel("y") 
    plt.title("LinearRegression(Regular Method)") 
    plt.legend() 
    plt.grid() 
    plt.show()
```

# GDM N FEATURES this is correct for n features and easiest GDM by me

```python

# Feature 0 (x0): Always an array of 1s to calculate the intercept (a0)
x0 = np.array([1, 1, 1, 1, 1, 1])

# Feature 1 (x1): Size of house (in sq ft)
x1 = np.array([15, 20, 12, 18, 24, 11])

# Feature 2 (x2): Number of bedrooms
x2 = np.array([3, 4, 2, 3, 4, 2])

# Target Variable (y): Price (in $1,000s)
y = np.array([250, 340, 185, 290, 410, 160])

x = np.array([
    x0,x1,x2
])

print(x,'\n\n',y)





def GDM(x,y,al,iters):

    a = np.array([0.0,0.0,0.0]) #ini a0 a1 a2
    
    
    m = len(y)

    learn = al/m

    for h in range(iters):
        
        y_c = a[0] + a[1]*x[1] + a[2]*x[2]

        error = y - y_c

        ta0 = a[0] - learn * np.sum(error * -x[0])
        ta1 = a[1] - learn * np.sum(error * -x[1])
        ta2 = a[2] - learn * np.sum(error * -x[2])

    
        a[0] = ta0
        a[1] = ta1
        a[2] = ta2

    return a


print(GDM(x,y,0.001,7000))
    
```
