```python
a = np.arange(1, 100)

result = np.array(list(filter(lambda x:x%3 == 0 or x%5 == 0,a)))

print("Multiples:", result)
print("Sum:", np.sum(result))
```

```python
a = np.array([1, 2, 3])
b = np.array([[4, 5, 6],
              [7, 8, 9]])

c = np.vstack((a, b))

print(c)
# Here a is 1D and b is 2D. vstack() combines them vertically.
```

```python

a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

c = np.dot(a, b)

# For matrix multiplication, use np.dot() or @.

print(c)
```

```python
# For the usual 3D cross product:

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.cross(a, b)

print(c)
```

```python
# 5. Write a NumPy program to compute the determinant of a given square array.
# 6. Write a NumPy program to compute the eigenvalues and right eigenvectors of a given
# square array.
# 7. Write a NumPy program to compute the inverse of a given matrix.
# 8. Write a NumPy program to compute the mean, standard deviation, and variance of a given
# array along the second axis.
```

```python
mat = np.array([
    [1,2,3],
    [4,5,6],
    [7,6,9],
])

print(np.linalg.det(mat))
```

```python
a = np.array([[1, 2],
              [2, 1]])

eigenvalues, eigenvectors = np.linalg.eig(a)

print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)
```

```python
mat = np.array([
    [1,2,3],
    [4,5,6],
    [7,6,9],
])

print(np.linalg.inv(mat))
```

```python

a = np.array([[1, 2, 3],
              [4, 5, 6]])

mean = np.mean(a, axis=1)
std = np.std(a, axis=1)
var = np.var(a, axis=1)

print("Mean:", mean)
print("Standard Deviation:", std)
print("Variance:", var)

# For a 2D NumPy array:
# axis=0 → column-wise operation
# axis=1 → row-wise operation
```
