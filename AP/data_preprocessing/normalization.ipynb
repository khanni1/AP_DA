```python
import pandas as pd 
data = {
   'a': [1,1,1,1,1], # zero var
   'b':  [2,3,2,3,2], # low var
   'c': [10,20,30,40,50], # high var
   'd':  [5,5,5,5,5], # zero
}

df = pd.DataFrame(data)
print(df)
```

```python
df.var()  # zero variance remove that cols/ feture A and D
```

```python
data = {
   'a': [1,1,1,1,1], # zero var
   'b':  [2,3,2,3,2], # low var
   'c': [10,20,30,40,50], # high var
   'd':  [5,5,5,5,5], # zero
}

from sklearn.feature_selection import VarianceThreshold

VT = VarianceThreshold(threshold=0.2) # threshold val then less then it is deleted

new = VT.fit_transform(df)

print(new)
```

```python
selected_columns = df.columns[VT.get_support()]
print(selected_columns)
```

```python
# Correlation Coefficient Method :
# keep features : highly correlatedd with target
#  remove feature : having very low correlation with targets
#  highly correlated with each other (both give same amount of information) remove that
```

```python
import pandas as pd 

data = {'Hours' : [2,4,6,8,10],'Attendance' : [60,65,70,80,90],'sleep' : [8,7,6,6,5],'marks' : [50,55,65,80,95]}

df = pd.DataFrame(data)
print(df)

#  here target is the marks outpyt and input us hours attendance and sleep
```

```python
co = df.corr()
print(co)
```

```python
'''
correlation with target = for relevance 
correlation between fetaures = for redudancy removal

note : multicollinearity (feature vs feature)
'''
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 

# heat map

sns.heatmap(co,annot=True , cmap = 'coolwarm')

plt.title("Correlation Heatmap")

plt.show()

```
