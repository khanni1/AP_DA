```python
# 1. imputing missing values
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = {'one' : pd.Series([1,2,5],index=['a','b','e']),'two':pd.Series([1,2,3,4],index=['a','b','c','d'])}

table = pd.DataFrame(data)

print(table)

# by default always convert to df any data
```

```python
table.isnull()
```

```python
table.isnull().any()
#  gives column name who have null values
```

```python
null_val = table.isnull().sum()
type(null_val)
```

```python
missing_pct = (null_val/len(table))*100
missing_pct

# how much of a col is not null
```

```python
# null handling
# using simple imputer

from sklearn.impute import SimpleImputer

si1 = SimpleImputer(missing_values = np.nan , strategy = 'mean')  # where ever nan replace with mean of that col
# this also for alias

# SHIFT + TAB FOR OFFLINE SUGGESTIONS

si2 = SimpleImputer(missing_values = np.nan , strategy = 'median')

si3 = SimpleImputer(missing_values = np.nan , strategy = 'most_frequent') # for mode 

si3.fit(table) 
temp = si3.transform(table)

# or

# si.fit_transform(table)


# now convert to dataframe
df1 = pd.DataFrame(temp,columns = ['one','two'])

df1
```

```python
# using fillna no needed sklearn here

x = table.fillna(20)
x
```

# 2.HANDLING CATEGORICAL DATA

## 2.1 Using get_dummies() (from pandas library)

```python
# categorical has textual data also we need to convert to numeric form

# can be nominal and ordinal 

''' nominal -> onehotencoder
           ->  get_dummies 
           
    oridnal -> labelencoder

    number of categories == number of columns created
'''



```

```python
data = {'customer_id' :pd.Series([1,2,3,4,5]),'loan_type' : pd.Series(['Home loan','Personal loan','Education loan','Car loan','Home loan']),'Income' : pd.Series(['15k','20k','30k','40k','50k'])}

print(type(data))

data = pd.DataFrame(data)

data
```

```python
pd.get_dummies(data['loan_type'])
loan_info = pd.get_dummies(data['loan_type'],dtype=int)

print(loan_info)

print()

income_info = pd.get_dummies(data['Income'],dtype=int)

print(income_info)
```

```python
whole_data = pd.get_dummies(data,dtype = int,prefix_sep = ' : ')
whole_data
```

### 2.2 using OneHotEncoder() (from sklearn library)

```python
from sklearn.preprocessing import OneHotEncoder

data = {'customer_id' :pd.Series([1,2,3,4,5]),'loan_type' : pd.Series(['Home loan','Personal loan','Education loan','Car loan','Home loan']),'Income' : pd.Series(['15k','20k','30k','40k','50k'])}

data = pd.DataFrame(data)

one = OneHotEncoder()

one
```

```python
x = one.fit_transform(data)
x1 = one.fit_transform(data).toarray()
print(x)
print()
print(x1)
```

```python
from sklearn.preprocessing import LabelEncoder

data = {'customer_id' :pd.Series([1,2,3,4,5]),'loan_type' : pd.Series(['Home loan','Personal loan','Education loan','Car loan','Home loan']),'Income' : pd.Series(['15k','20k','30k','40k','50k'])}

data = pd.DataFrame(data)

le = LabelEncoder()
le

```

# 3. Scaling of Data

### 3.1 MinMax Scaler
### 3.2 Standard Scaler
### 3.3 Robust Scaler

```python
'''
SCALING IS TO MAKE TO RANGE OF DATA BETWEEN A CERTAIN RANGE OF WHOLE DATASET

3.1 minmax 
x = (x - xmin) / (xmax - xmin)

3.2 Standard scaler
x = (x - μ) /  σ 

3.3 Robust Scaler
x = (x - median) / IQR

'''
```

```python
data = np.array([[-100.3],[27.5],[0],[-200],[1000]])



print('before scaling \n',data,'\n')

```

```python
from sklearn import preprocessing

mm = preprocessing.MinMaxScaler()
mm1 = preprocessing.MinMaxScaler(feature_range=(5,10))
mm2 = preprocessing.MinMaxScaler(feature_range=(100,200))
```

```python
scaled = mm2.fit_transform(data)

scaled
```

```python
ss = preprocessing.StandardScaler()
```

```python
x = ss.fit_transform(data)
x
# -3 to 3 range here
```

```python
rs = preprocessing.RobustScaler()

x = rs.fit_transform(data)
x
```
