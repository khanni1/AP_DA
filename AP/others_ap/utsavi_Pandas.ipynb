### Creation of Series

#### (A)Creation of Series from Scalar Values

```python
# series and dataframe in pandas
# default 0 index series
# series (lst,index) 
# hetergenous index allowed in series
```

```python
import pandas as pd
import numpy as np
```

```python
series1 = pd.Series([10,20,30])
print(series1)
print(type(series1))
```

### Mathematical Operations on Series

```python
seriesA = pd.Series([1,2,3,4,5], index =['a', 'b', 'c', 'd', 'e'])
```

```python
seriesA
```

```python
seriesB = pd.Series([10,20,-10,-50,100],index = ['z', 'y', 'a', 'c', 'e'])
```

```python
seriesB
```

#### (A)Addition of two Series

```python
seriesA + seriesB
```

```python
seriesA.add(seriesB, fill_value=0.0)
```

#### (B)Subtraction of two Series

```python
seriesA - seriesB
```

```python
seriesA.sub(seriesB, fill_value=1000)
```

#### (C)Multiplication of two Series

```python
seriesA * seriesB
```

```python
seriesA.mul(seriesB, fill_value=0)
```

#### (D)Division of two Series

```python
seriesA/seriesB
```

### Joining, Merging and Concatenation of DataFrames

#### (A)Joining

```python
dFrame1=pd.DataFrame([[1, 2, 3], [4, 5],[6]], columns=['C1', 'C2', 'C3'], index=['R1','R2', 'R3'])
dFrame1
```

```python
dFrame2=pd.DataFrame([[10, 20], [30], [40,50]], columns=['C2', 'C5'], index=['R4', 'R2','R5'])
dFrame2
```

```python
dFrame1=dFrame1.append(dFrame2)
dFrame1
```

```python
dFrame2 =dFrame2.append(dFrame1,sort='True')
dFrame2
```

```python
dFrame2 = dFrame2.append(dFrame1,sort=’False’)
dFrame2
```
