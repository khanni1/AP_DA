# HYPOTHESIS TESTING 
### SIGNMA KNOWN CASE

```python
# LOWER tailed z test

import numpy as np
from scipy.stats import norm

# sample data
xbar = 48
hypothesied_value = 50
sig = 5
n = 30
al = 0.05

# calcualte test statistic
z_cal = (xbar - hypothesied_value)/(sig / np.sqrt(n))

# calculate p-valueabs
p_value = norm.cdf(z_cal)
# use z table using this norm.cdf


# DECISION 

print(f"Lower-tailed Z test Result")

if p_value < al:
    print("Reject the null hypothesis")
else:
    print("failed to reject null hypothesis")

print(z_cal)
print(p_value)
```

```python
# UPPER tailed z test

# sample data
xbar = 52
hypothesied_value = 50
sig = 5
n = 30
al = 0.05

# calcualte test statistic
z_cal = (xbar - hypothesied_value)/(sig / np.sqrt(n))

# calculate p-valueabs
p_value = 1 - norm.cdf(z_cal)


# DECISION 

print(f"Lower-tailed Z test Result")

if p_value < al:
    print("Reject the null hypothesis")
else:
    print("failed to reject null hypothesis")

print(z_cal)
print(p_value)
```

```python
# TWO tailed z test

# sample data
xbar = 52
hypothesied_value = 50
sig = 5
n = 30
al = 0.05/2

# calcualte test statistic
z_cal = (xbar - hypothesied_value)/(sig / np.sqrt(n))

# calculate p-value
if z_cal < 0:
    
    p_value = 1 - norm.cdf(abs(z_cal))
else:
    p_value = norm.cdf(abs(z_cal))
    

# DECISION 

print(f"Lower-tailed Z test Result")

if p_value < al:
    print("Reject the null hypothesis")
else:
    print("failed to reject null hypothesis")

print(z_cal)
print(p_value)
```

### SIGMA UNKOWN CASE

```python
# lower tail 

sample = np.array([48,49,47,46,49,48,47,46,48,47])

hypothesied_value = 50

al = 0.05

# Calculate sample statistics
s = np.std(sample,ddof=1)  #because in deno n-1 is degree of freedom
xbar = np.mean(sample)
n = len(sample)

t_cal = (xbar - hypothesied_value)/(s/np.sqrt(n))

# calcalte p-value 
p_value = t.cdf(t_cal,df=n-1)

print(p_value)
print(t_cal)

if p_value < al:
    print("Reject the null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

```python
# upper tail 

sample = np.array([48,49,47,46,49,48,47,46,48,47])

hypothesied_value = 50

al = 0.05

# Calculate sample statistics
s = np.std(sample,ddof=1)  #because in deno n-1 is degree of freedom
xbar = np.mean(sample)
n = len(sample)

t_cal = (xbar - hypothesied_value)/(s/np.sqrt(n))

# calcalte p-value 
p_value = 1 - t.cdf(t_cal,df=n-1)

print(p_value)
print(t_cal)

if p_value < al:
    print("Reject the null hypothesis")
else:
    print("Fail to reject null hypothesis")
```
