```python
import math

# helper: mathematical position -> Python index def at(pos): return x[pos - 1]
# very imp ! memorise by heart for future conversions 
def at(pos, x): 
    n = len(x) 
    if pos < 1: 
        pos = 1 
    elif pos > n: 
        pos = n 
    return x[pos - 1]
```

```python
data = [10, 20, 30, 40]

data = sorted(data)

n = len(data)

if n % 2 == 1:
    median = data[n // 2]  #odd number of observations
else:
    median = (data[n // 2 - 1] + data[n // 2]) / 2

print(median)
```

```python
data = [10, 20, 20, 30, 20, 40]

frequency = {}

for x in data:
    if x in frequency:
        frequency[x] = frequency[x] + 1
    else:
        frequency[x] = 1

print(frequency)
```

```python

def percentile(p, x: list):
    x = sorted(x)
    n = len(x)

    v1 = (p / 100) * n

    if v1 % 1 != 0:
        # decimal position → round UP
        pos = Math.ceil(v1)
        return at(pos, x)

    else:
        # integer position
        v1 = int(v1)

        i1 = v1
        i2 = v1 + 1

        ans = (at(i1, x) + at(i2, x)) / 2
        return ans
```

# PROBABILITY CONCEPTS

```python
import random

heads = 0
trials = 1000

for i in range(trials):
    result = random.choice(["H", "T"])

    if result == "H":
        heads += 1

probability = heads / trials

print(probability)

'''
It won't necessarily be exactly 0.5 because this is a random experiment.

Viva

Why isn't the result exactly 0.5?

Because we're estimating theoretical probability using a finite number of random trials.
'''

```
