```python
# helper: mathematical position -> Python index def at(pos): return x[pos - 1]
# very imp ! memorise by heart for future conversions 
def at(pos, x): 
    n = len(x) 
    if pos < 1: 
        pos = 1 
    elif pos > n: 
        pos = n 
    return x[pos - 1]
```

```python
def mean(x:list):
    return (sum(x)/len(x))
```

```python
print(mean(x))
```

```python
def median(x:list):
    x = sorted(x)

    if(len(x) % 2 == 1):
        mi = (len(x) // 2)
        return x[mi]
    else :
        mi2 = len(x) // 2
        mi1 = mi2-1
        

        avg = (x[mi1] + x[mi2])/2

        return avg
    
```

```python
def mode1(x:list):
    x = sorted(x)

    count = {}

    for i in x:
        if i in count:
            count[i] += 1
        else:
            count[i] = 1

    max1 = 0

    for i in count:
        if(count[i] > max1):
            max1 = count[i]
    modes = []
    for k,v in count.items():
        if(v == max1):
            modes.append(k)

    return modes
            

    
def mode2(x:list):
    ans = max(set(x),key=x.count)
    return ans


# so first arr = [1,1,2,2,3]
# set makes it [1,2,3]
# then for each value in that set as s
# arr.count(s)
# is done and then from the output list max is selected.

def mode3(x:list):
    countx = {}
    s = sorted(set(x))
    for i in s:
        countx[i] = x.count(i)

    max_freq = 0
    for k,v in countx.items():
        if(v > max_freq):
            max_freq = v
            
    modes = []
    for k,v in countx.items():
        if(v == max_freq):
            modes.append(k)
    return modes            
        
print(mode1(x))
print(mode2(x))
print(mode3(x))
       
    
```

```python
def percentile(p,x:list):
    x = sorted(x)
    n = len(x)

    v1 = (p/100)*n

    if(v1 % 1 != 0):
        # means its some point (decimal) value
        pos = int(v1)
        
        return at(pos,x)
    else :
        # if v1 is integer
        # enforced interger if 7.0 => 7
        v1 = int(v1)
        
        i1 = v1
        i2 = v1+1
        
        ans = (at(i1,x) + at(i2,x)) / 2
        return ans

print(percentile(0,y))
        
```

```python
def IQR(x:list):
    Q1 = percentile(25,x)
    Q3 = percentile(75,x)

    return (Q3 - Q1)

print(IQR(y))
```

```python
def variance(x:list):
    m = len(x)
    xm = mean(x)
    summ = 0
    
    for i in x:
        temp = i - xm
        temp = temp**2
        summ = summ + temp

    var = summ/(m-1)

    return var
        
    
```

```python
def standard_dev(x:list):
    var = variance(x)
    return var**0.5
```
